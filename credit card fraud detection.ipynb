# Credit card fraud detection

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score,confusion_matrix
```

```python
# Loading the credit_card_data:
cd_data = pd.read_csv("creditcard.csv")
cd_data
```

```python
cd_data.info()
```

```python
# Chechking that data_set has null values:
cd_data.isnull().sum()
```

```python
cd_data.describe()
```

```python
# legit transcation or non_legit detail in class columns:
# 1 represent the non_legit transactions:
# 0 represents the legit transactions:
cd_data['Class'].value_counts()
```

```python
# Separating the legit and non_legit transcations:
legit = cd_data[cd_data.Class==0]
non_legit = cd_data[cd_data.Class==1]
pd.DataFrame(legit)
```

```python
pd.DataFrame(non_legit)
```

```python
print(legit.shape) # non_Fraudulent data:
print(non_legit.shape) # Fraudulent data 
```

```python
# Stastical measures of the legit and non_legit data: 
ls = legit.Amount.describe()
ls
```

```python
n_ls = non_legit.Amount.describe()
print(n_ls)
```

```python
#comparing legit and non_legit on the basis of their mean value:
comp = cd_data.groupby('Class').mean()
comp
```

```python
# UNDER-SAMPLING:
# Taking the n no.of values as of non_legit data from legit data:
# The data is unbalanced so we have done the under_sampling. 

legit_sample = legit.sample(n=492)
legit_sample
```

```python
new_ds = pd.concat([legit_sample,non_legit],axis=0)
new_ds.head()
```

```python
new_ds.tail()
```

```python
new_ds['Class'].value_counts()
```

```python
new_ds.groupby("Class").mean()
```

```python
# Splitting the data into features and target values:

X = new_ds.drop(columns='Class',axis=1)
Y = new_ds['Class']
```

```python
X
X.shape
print(X)
```

```python
Y
```

```python
X_train,X_test,Y_train,Y_test = train_test_split(X,Y,test_size = 0.2,stratify = Y,random_state=2)
print(X.shape,X_train.shape,X_test.shape)
print(Y_test.shape,Y_train.shape)
```

```python
model = LogisticRegression()
```

```python
model.fit(X_train,Y_train)
```

```python
# testing the model and its accuracy: 

X_train_prediction = model.predict(X_train)

training_data_accuracy = accuracy_score(X_train_prediction,Y_train)

print(f"Accuracy of X_train: {training_data_accuracy} ")
```

```python
print(f"Training_data_prediction: {X_train_prediction} ")
```

```python
# Now Fitting the test value:

model.fit(X_test,Y_test)
```

```python
# predicting the test value and checking their accuracy score: 

X_test_prediction = model.predict(X_test)
X_test_accuracy = accuracy_score(X_test_prediction,Y_test)
```

```python
# printing the of accuracy of the test values: 
print(f"Acuuracy of X_test: {X_test_accuracy} ")
```

```python
print(X_test_prediction)
```

```python
c_matrix = confusion_matrix(X_test_prediction,Y_test)
c_matrix
```

```python
predicted_data = pd.DataFrame(np.c_[X_test,Y_test,X_test_prediction])
predicted_data
```

```python
# --- Simple Command-Line Interface Starts Here ---

print("\n--- Credit Card Fraud Detection Test Interface ---")
print("Enter transaction details below. Type 'quit' at any prompt to exit.")

# Get the feature names in the order the model expects them
feature_names = X.columns.tolist()

while True:
    input_data_list = []
    try:
        print("\nPlease enter the following transaction features:")
        for feature in feature_names:
            user_input = input(f"  Enter value for {feature}: ")
            if user_input.lower() == 'quit':
                raise KeyboardInterrupt # Use an exception to break out cleanly

            # Attempt to convert input to float
            input_data_list.append(float(user_input))

        # Convert list to NumPy array and reshape for prediction (1 row, 30 columns)
        input_data_array = np.array(input_data_list).reshape(1, -1)

        # Make prediction
        prediction = model.predict(input_data_array)
        prediction_proba = model.predict_proba(input_data_array) # Get probabilities

        # Display result
        print("\n--- Prediction Result ---")
        if prediction[0] == 0:
            print("  Prediction: Legitimate Transaction")
        else:
            print("  Prediction: >> Fraudulent Transaction <<")

        print(f"  Probability -> [Legitimate: {prediction_proba[0][0]:.4f}, Fraudulent: {prediction_proba[0][1]:.4f}]")
        print("-------------------------\n")

    except ValueError:
        # Handle cases where input is not a valid number
        print("\nError: Invalid input. Please enter numeric values only.")
        print("Restarting transaction input...\n")
        continue # Skip rest of the loop and start asking for input again

    except KeyboardInterrupt:
        # Handle user typing 'quit'
        print("\nExiting test interface.")
        break # Exit the while loop

    except Exception as e:
        # Catch any other unexpected errors
        print(f"\nAn unexpected error occurred: {e}")
        print("Exiting test interface.")
        break

print("--- Interface Closed ---")
```
